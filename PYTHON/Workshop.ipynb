```python
import cv2
import numpy as np
from IPython.display import Image, display
```

```python
# Read the original image
filename = "sample_image.jpg"
image = cv2.imread(filename)

# Check if image is loaded properly
if image is None:
    print(f"Error: Could not load image from {filename}. Make sure the file exists.")
else:
    print("Original image loaded successfully.")
    display(Image(filename))
```

```python
# Resize image
if image is not None:
    resized = cv2.resize(image, (400, 300))
    cv2.imwrite("resized_output.jpg", resized)
    print("Resized image saved successfully.")
    display(Image("resized_output.jpg"))
```

```python
# Convert to grayscale
if image is not None:
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    cv2.imwrite("gray_output.jpg", gray_image)
    print("Grayscale image saved successfully.")
    display(Image("gray_output.jpg"))
```

```python
# Gaussian Blur
if image is not None:
    blurred = cv2.GaussianBlur(image, (15, 15), 0)
    cv2.imwrite("blur_output.jpg", blurred)
    print("Blurred image saved successfully.")
    display(Image("blur_output.jpg"))
```

```python
# Edge Detection (Canny)
if image is not None:
    edges = cv2.Canny(image, 100, 200)
    cv2.imwrite("edges_output.jpg", edges)
    print("Edge detection image saved successfully.")
    display(Image("edges_output.jpg"))
```

```python
# Thresholding
if image is not None:
    _, binary = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
    cv2.imwrite("binary_output.jpg", binary)
    print("Binary threshold image saved successfully.")
    display(Image("binary_output.jpg"))
```

```python
# Rotation
if image is not None:
    (h, w) = image.shape[:2]
    center = (w // 2, h // 2)
    M = cv2.getRotationMatrix2D(center, 45, 1.0)
    rotated = cv2.warpAffine(image, M, (w, h))
    cv2.imwrite("rotated_output.jpg", rotated)
    print("Rotated image saved successfully.")
    display(Image("rotated_output.jpg"))
```

```python
# Cropping
if image is not None:
    cropped = image[100:400, 200:600]
    cv2.imwrite("cropped_output.jpg", cropped)

    print("Cropped image saved successfully.")
    display(Image("cropped_output.jpg"))
```
